```python
nama_barang = "Sapu"
harga_barang = 15000
jumlah_beli = 3

total_beli = harga_barang * jumlah_beli
total_diskon = total_beli * 10 / 100
total_pembayaran = total_beli - total_diskon

print(f"Nama Barang: {(nama_barang)}")
print(f"Total: Rp {int(total_beli)}")
print(f"Diskon: Rp {int(total_beli*10/100)}")
print(f"Bayar: Rp {int(total_pembayaran)}")
```
