```python
MULAI

 Tampilkan judul "Sistem Pembayaran Parkir Kampus"

 Siapa nama pemilik kendaraan?
 Apa statusnya? Mahasiswa, Dosen, atau Pengunjung?
 Kendaraannya apa? Motor atau Mobil?
 Sudah parkir berapa jam?

 Jika Motor, harga jam pertama 2000 rupiah,
   jam berikutnya tambah 1000 rupiah per jam.
 Jika Mobil, harga jam pertama 5000 rupiah,
   jam berikutnya tambah 2000 rupiah per jam.
 Jika selain Motor dan Mobil, tampilkan pesan error
   dan program berhenti.

 Jika parkirnya lebih dari 1 jam,
   biaya tambahan = (lama parkir - 1 jam) dikali harga per jam tambahan.
 Jika cuma 1 jam atau kurang, tidak ada biaya tambahan.

 Jumlahkan harga awal + biaya tambahan,
   hasilnya jadi total sementara.

 Jika Mahasiswa, dapat diskon 20%
 Jika Dosen, dapat diskon 30%
 Jika Pengunjung, tidak dapat diskon
 Jika statusnya tidak dikenal, anggap sebagai Pengunjung
   dan tampilkan peringatan.

 diskon = total sementara dikali persentase diskon

 total bayar = total sementara dikurangi diskon

 Nama pemilik
 Status dan berapa persen diskonnya
 Jenis kendaraan dan lama parkirnya
 Biaya awal
 Biaya tambahan
 Jumlah diskon
 Total yang harus dibayar

SELESAI
```

```python
def hitung_parkir():
    print("=== Sistem Pembayaran Parkir Kampus ===")

    nama_pemilik = input("Masukkan Nama Pemilik: ")
    status_pengguna = input("Status Pengguna (Mahasiswa/Dosen/Pengunjung): ").strip().capitalize()
    jenis_kendaraan = input("Jenis Kendaraan (Motor/Mobil): ").strip().capitalize()
    lama_parkir = int(input("Lama Parkir (jam): "))

    if jenis_kendaraan == "Motor":
        tarif_awal = 2000
        tarif_tambahan_per_jam = 1000
    elif jenis_kendaraan == "Mobil":
        tarif_awal = 5000
        tarif_tambahan_per_jam = 2000
    else:
        print("\nError: Jenis kendaraan tidak valid. Silakan masukkan 'Motor' atau 'Mobil'.")
        return

    if lama_parkir > 1:
        biaya_tambahan = (lama_parkir - 1) * tarif_tambahan_per_jam
    else:
        biaya_tambahan = 0

    total_sementara = tarif_awal + biaya_tambahan

    if status_pengguna == "Mahasiswa":
        persentase_diskon = 0.20
    elif status_pengguna == "Dosen":
        persentase_diskon = 0.30
    elif status_pengguna == "Pengunjung":
        persentase_diskon = 0.00
    else:
        print("\nPeringatan: Status tidak dikenali. Dianggap sebagai 'Pengunjung' (Diskon 0%).")
        persentase_diskon = 0.00

    nominal_diskon = total_sementara * persentase_diskon
    total_pembayaran = total_sementara - nominal_diskon

    print("\n--- Struk Pembayaran Parkir ---")
    print(f"Nama Pemilik     : {nama_pemilik}")
    print(f"Status           : {status_pengguna} (Diskon {int(persentase_diskon * 100)}%)")
    print(f"Kendaraan        : {jenis_kendaraan} ({lama_parkir} Jam)")
    print("-" * 33)
    print(f"Biaya Awal       : Rp {tarif_awal:,.0f}".replace(',', '.'))
    print(f"Biaya Tambahan   : Rp {biaya_tambahan:,.0f}".replace(',', '.'))
    print(f"Diskon           : Rp {nominal_diskon:,.0f}".replace(',', '.'))
    print(f"Total Pembayaran : Rp {total_pembayaran:,.0f}".replace(',', '.'))

if __name__ == "__main__":
    hitung_parkir()
```
