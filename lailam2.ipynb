```python
nilai = int(input("masukkan nilai:"))

if nilai >= 90:
            grade = "A"

elif nilai >= 80:
            grade = "B"

elif nilai >= 75:
            grade = "C"

elif nilai >= 70:
            grade = "D"

elif nilai >= 65:
            grade = "E"
else:
            grade = "tidak lulus"
print("nilai:", nilai)
print("grade:", grade)
```

```python
# Tampilkan menu restoran
print("=== MENU RESTORAN ===")
print("1. Nasi Goreng (Rp 15.000)")
print("2. Mie Ayam (Rp 12.000)")
print("3. Es Teh (Rp 3.000)")

# Minta input pilihan dan jumlah
pilihan = int(input("Masukkan pilihan menu (1-3): "))
jumlah = int(input("Masukkan jumlah pesanan: "))

# Tentukan harga menggunakan if-elif-else (pengganti switch-case)
if pilihan == 1:
    harga = 15000
    nama_menu = "Nasi Goreng"
    total = harga * jumlah
    print(f"Total bayar: Rp {total}")
elif pilihan == 2:
    harga = 12000
    nama_menu = "Mie Ayam"
    total = harga * jumlah
    print(f"Total bayar: Rp {total}")
elif pilihan == 3:
    harga = 3000
    nama_menu = "Es Teh"
    total = harga * jumlah
    print(f"Total bayar: Rp {total}")
else:
    # Default case jika pilihan tidak valid
    print("Menu tidak ada")
```
